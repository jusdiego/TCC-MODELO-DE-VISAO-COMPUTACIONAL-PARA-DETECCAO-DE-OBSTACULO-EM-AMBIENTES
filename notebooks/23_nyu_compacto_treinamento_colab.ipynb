## 2. Recuperar a base e a execução compatível

Só o ZIP de desenvolvimento é recuperado nesta etapa. O YAML de treino
contém TRAIN e VAL; nenhum caminho de TEST é entregue ao treinador.
Checkpoints e histórico são conferidos por SHA-256. O dispositivo pode
mudar entre sessões; o ambiente de cada sessão é registrado e não se
promete igualdade bit a bit com um treino ininterrupto.

Primeiro procura os manifestos nas pastas próprias desta preparação, incluindo
uma pasta `nyu22_local` enviada ao Drive. Não varre fotos, bases antigas ou treinos.
Reconhece a estrutura original e a preparação em `resultados` quando os dois
ZIPs e o manifesto estão juntos. Usa a localização existente dos arquivos;
checkpoints continuam na pasta canônica deste ramo. Não escolhe entre bases
diferentes apenas pela data e não substitui uma referência canônica por outra base.


```python
try:
    STORAGE,LOCALIZACAO=locator.resolve(STORAGE,LOCAL/'localizacao_preparacao')
    JOB=workflow.session(STORAGE,LOCAL,BUNDLE,device=DEVICE)
    OUTPUT=JOB['output'];REMOTE=JOB['remote']
    LOCALIZACAO['locator_sha256']=LOCALIZADOR_SHA256
    c.save(OUTPUT/'localizacao_preparacao_nyu23.json',LOCALIZACAO)
    shutil.copytree(BOOT/'codigo',OUTPUT/'codigo',dirs_exist_ok=True)
    def backup():return training.sync(STORAGE,OUTPUT,REMOTE)
    print('Imagens de treino:',len(JOB['train']),'| validação:',len(JOB['val']))
    ANTERIOR=training.local_resume(OUTPUT,JOB['context'])
    print('Épocas completas recuperadas:',ANTERIOR[0]['epoch'] if ANTERIOR else 0)
    backup()
except Exception as exc:
    BUSCA=LOCAL/'localizacao_preparacao'/'busca_preparacao_nyu23.json'
    if BUSCA.is_file():shutil.copy2(BUSCA,OUTPUT/BUSCA.name)
    falha(exc,'23-base-e-retomada');raise
```
